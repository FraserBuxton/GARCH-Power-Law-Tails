```python
import numpy as np
from scipy import stats
```

# Proper Scoring
- Compute Log-Score for Results
- Compute CRPS for Results
- Test Significance with DM Test

## Log-Score

```python
def log_score(val, mu, sigma):
    '''Compute Log-Score'''
    return 0.5*np.log(2*np.pi*sigma**2) + (val - mu)**2 / (2*sigma**2)
```

## CRPS

```python
def crps(y, mu, sigma):
    '''Compute CRPS'''
    z = (y - mu) / sigma
    return sigma * (z * (2 * stats.norm.cdf(z) - 1) +
                    2 * stats.norm.pdf(z) -
                    1 / np.sqrt(np.pi))
```

## DM Tests

```python
def dm_test(loss1, loss2):
    '''Compute DM Test Statistic from Losses'''

    d = loss2 - loss1
    T = len(d)

    d_bar = np.mean(d)
    var = np.var(d, ddof=1)
    
    dm_stat = d_bar / np.sqrt(var / T)
    p_value_plus = stats.norm.cdf(dm_stat) # Model 1 Better
    p_value_minus = 1 - stats.norm.cdf(dm_stat) # Model 2 Better

    if (p_value_minus <= 0.05):
        print('Model 1 Better')
    elif (p_value_plus <= 0.05):
        print('Model 2 Better')
    else:
        print('Inconclusive')
    return dm_stat, p_value_plus, p_value_minus
```
