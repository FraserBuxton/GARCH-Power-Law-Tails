# Tail Estimation
- Estimate Tail Index Through Hill Estimator
- Estimate Tail Index Through Moment Estimator

## Hill Estimator

```python
def hill_estimator(ret, k):
    '''Hill Estimator for Tail Index'''
    ret = np.flip(np.sort(ret))
    
    X = ret[k]
    y = ret[:k]
    
    gamma = np.mean(np.log(y) - np.log(X))
    return 1/gamma
```

```python
def smooth(ts, u):
    '''Smooth a Time-Series'''
    smoothed = np.empty(int(len(ts)/u))
    
    for i in range(1, len(smoothed)):
        smoothed[i] = np.mean(ts[i+1:int(u*i+1)])
        
    return smoothed
```

```python
def hill_k(hills, window):
   '''Find Optimal k for Hill Estimate'''
        
   vars = np.zeros([len(hills) - window*2])
   for j in range(len(vars)):
      vars[j] = np.var(hills[j:j + window*2])
       
   k = np.argmin(vars)+window
       
   return vars, k, hills[k]
```

## Moment Estimator

```python
def moment_estimator(ret, k):
    '''Moment Estimator for Tail Index'''
    ret = np.flip(np.sort(ret))
    
    X = ret[k]
    y = ret[:k]
    
    excess = np.log(y) - np.log(X)
    M1 = np.mean(excess)
    M2 = np.mean(excess**2)
    
    gamma = M1 + 1 - 0.5 * (1 - (M1**2 / M2))**(-1)
    return 1/gamma
```

```python
def moment_k(moments, window):
   '''Optimise k for Moment Estimate'''
        
   vars = np.zeros([len(moments) - window*2])
   for j in range(len(vars)):
       vars[j] = np.var(moments[j:j + window*2])
       
   k = np.argmin(vars)+window
       
   return vars, k, moments[k]
```
