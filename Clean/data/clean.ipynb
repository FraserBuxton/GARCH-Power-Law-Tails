```python
import pandas as pd
import scipy.stats as stats
import numpy as np
```

```python
data = pd.read_pickle('forced_ti361_data.pk')
data
```

```python
def crps(y, mu, sigma):
    z = (y - mu) / sigma
    return sigma * (z * (2 * stats.norm.cdf(z) - 1) +
                    2 * stats.norm.pdf(z) -
                    1 / np.sqrt(np.pi))
```

```python
data['CRPS'] = crps(data['Returns'], data['Mean'], data['Standard Deviation'])
```

```python
def log_score(val, mu, sigma):
    return 0.5*np.log(2*np.pi*sigma**2) + (val - mu)**2 / (2*sigma**2)
```

```python
data['Log-Score'] = log_score(data['Returns'], data['Mean'], data['Standard Deviation'])
```

```python
pd.to_pickle(data, 'forced_ti324_data.pk')
```
